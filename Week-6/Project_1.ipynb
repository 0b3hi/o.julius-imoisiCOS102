```python
import tkinter as tk
from tkinter import messagebox
import random

# Data
employees = ["Mary Evans", "Eyo Ishan", "Durojaiye Dare", "Adams Ali",
             "Andrew Ugwu", "Stella Mankinde", "Jane Akibo",
             "Ago James", "Michell Taiwo", "Abraham Jones",
             "Nicole Anide", "Kosi Korso", "Adele Martins",
             "Emmanuel Ojo", "Ajayi Fatima"]

tasks = ["Loading", "Transporting", "Reviewing Orders",
         "Customer Service", "Delivering Items"]


# OOP CLASS
class Employee:
    def __init__(self, name):
        self.name = name

    def check_employee(self):
        return self.name in employees

    def take_attendance(self):
        return f"Attendance taken for {self.name}"

    def assign_task(self):
        task = random.choice(tasks)
        return f"Task assigned: {task}"

    def refuse_access(self):
        return "Access Denied! You are not an employee."


# FUNCTION FOR BUTTON
def submit():
    name = entry.get()
    emp = Employee(name)

    if emp.check_employee():
        attendance = emp.take_attendance()
        task = emp.assign_task()
        messagebox.showinfo("Success", f"{attendance}\n{task}")
    else:
        messagebox.showerror("Error", emp.refuse_access())


# GUI
root = tk.Tk()
root.title("Employee System")
root.geometry("400x250")

label = tk.Label(root, text="Enter Employee Name", font=("Arial", 14))
label.pack(pady=10)

entry = tk.Entry(root, width=30)
entry.pack(pady=10)

btn = tk.Button(root, text="Submit", command=submit)
btn.pack(pady=10)

root.mainloop()
```
