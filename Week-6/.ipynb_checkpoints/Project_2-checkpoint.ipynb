```python
import tkinter as tk
from tkinter import messagebox

# OOP CLASS
class DeliveryService:
    def __init__(self, location, weight):
        self.location = location.lower()
        self.weight = weight

    def calculate_fee(self):
        if self.location == "pau":
            if self.weight >= 10:
                return 2000
            else:
                return 1500

        elif self.location == "epe":
            if self.weight >= 10:
                return 5000
            else:
                return 4000

        else:
            return None


# FUNCTION FOR BUTTON
def calculate():
    location = entry_location.get()
    
    try:
        weight = float(entry_weight.get())
    except ValueError:
        messagebox.showerror("Error", "Please enter a valid weight")
        return

    delivery = DeliveryService(location, weight)
    fee = delivery.calculate_fee()

    if fee is None:
        messagebox.showerror("Error", "Invalid location! Use PAU or Epe.")
    else:
        messagebox.showinfo(
            "Delivery Fee",
            f"Location: {location}\nWeight: {weight}kg\nFee: ₦{fee}"
        )


# GUI SETUP
root = tk.Tk()
root.title("Delivery Fee Calculator")
root.geometry("400x300")

# Location
label1 = tk.Label(root, text="Enter Location (PAU or Epe):", font=("Arial", 12))
label1.pack(pady=10)

entry_location = tk.Entry(root, width=30)
entry_location.pack(pady=5)

# Weight
label2 = tk.Label(root, text="Enter Weight (kg):", font=("Arial", 12))
label2.pack(pady=10)

entry_weight = tk.Entry(root, width=30)
entry_weight.pack(pady=5)

# Button
btn = tk.Button(root, text="Calculate Fee", command=calculate)
btn.pack(pady=20)

root.mainloop()
```
