```python
import tkinter as tk
import pandas as pd

# load dataset
data = pd.read_csv("GIG-logistics.csv")

def check_employee():
    name = name_entry.get()
    dept = dept_entry.get()

    result = data[(data["Name"] == name) & (data["Department"] == dept)]

    if not result.empty:
        same_dept = data[data["Department"] == dept]["Name"].tolist()

        message = "Welcome " + name + "\n\nOther employees in " + dept + ":\n"
        message += "\n".join(same_dept)

        result_label.config(text=message)

    else:
        result_label.config(text="Employee does not exist.")

# GUI window
window = tk.Tk()
window.title("Employee Verification")
window.geometry("600x400")

tk.Label(window, text="Enter Name").pack()
name_entry = tk.Entry(window)
name_entry.pack()

tk.Label(window, text="Enter Department").pack()
dept_entry = tk.Entry(window)
dept_entry.pack()

tk.Button(window, text="Check Employee", command=check_employee).pack()

result_label = tk.Label(window, text="")
result_label.pack()

window.mainloop()
```
